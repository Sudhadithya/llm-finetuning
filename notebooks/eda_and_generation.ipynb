# Customer Support Dataset Exploration

This notebook demonstrates how to load, visualize, and analyze the domain-specific data we fine-tune our models on.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
# Load data
df_train = pd.read_json('../datasets/customer_support/train.jsonl', lines=True)
print(f"Training samples: {len(df_train)}")
df_train.head()
```

```python
# Analyze Sequence Lengths
df_train['prompt_len'] = df_train['prompt'].apply(lambda x: len(x.split()))
df_train['response_len'] = df_train['response'].apply(lambda x: len(x.split()))

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df_train['prompt_len'], bins=20, ax=ax[0]).set_title('Prompt Token Lengths')
sns.histplot(df_train['response_len'], bins=20, ax=ax[1]).set_title('Response Token Lengths')
plt.show()
```
